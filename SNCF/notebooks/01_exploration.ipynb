# Exploration - Régularité mensuelle TGV

```python
import pandas as pd
pd.set_option("display.max_columns", None)

df = pd.read_csv("../data/raw/regularite_tgv.csv", sep=";")
df.shape
```

```python
df.info()
df.head()
```

```python
print(df.isna().sum())
print("Doublons :", df.duplicated().sum())
```

```python
df.describe().T
```

```python
num = df.select_dtypes("number")
print("Valeurs négatives :\n", (num < 0).sum()[(num < 0).sum() > 0])

print("Trains prévus = 0 :", (df.nb_train_prevu == 0).sum())
print("Annulés > prévus :", (df.nb_annulation > df.nb_train_prevu).sum())
print("Durée trajet = 0 :", (df.duree_moyenne == 0).sum())

causes = [c for c in df.columns if c.startswith("prct_")]
print("Somme des causes ≠ 100 :", (df[causes].sum(axis=1).round() != 100).sum())
```

```python
print("Gares :", df.gare_depart.nunique())
print("Liaisons :", (df.gare_depart + " → " + df.gare_arrivee).nunique())

annee = df.groupby(df.date.str[:4])[["nb_train_retard_arrivee", "nb_train_prevu", "nb_annulation"]].sum()
annee["ponctualite"] = 100 - annee.nb_train_retard_arrivee / (annee.nb_train_prevu - annee.nb_annulation) * 100
annee["ponctualite"].round(1)
```
